```python
import csv

with open("emp.csv", 'w', newline='') as f:
    w = csv.writer(f)
    w.writerow(["ENO", "ENAME", "ESALARY", "EADDR"])

    n = int(input("Enter number of employees : "))

    for i in range(n):
        eno = input("Enter employee no : ")
        ename = input("Enter employee name : ")
        esal = input("Enter employee salary : ")
        eaddr = input("Enter employee address : ")

        w.writerow([eno, ename, esal, eaddr])

print("Data inserted successfully!")
```

```python
import numpy as np

students = np.array([
    [85, 90, 78],  # Student 1
    [60, 55, 70],  # Student 2
    [95, 92, 88],  # Student 3
    [40, 50, 45],  # Student 4
    [72, 68, 75]   # Student 5
])

students
```

```python
avg_marks = np.mean(students, axis=1)

print("Average Marks :", avg_marks)
```

```python
topper = np.argmax(avg_marks)

print("Topper Student:", topper)
```

```python
# Performance Report

for i, avg in enumerate(avg_marks):
    status = "Pass" if avg > 50 else "Fail"
    print(f"Student {i+1}: Avg = {avg:.2f} | {status}")
```

```python
x = [1, 2, 3, 4]
y = [2, 4, 6, 8]

plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel("X axis")
plt.ylabel("Y Axis")
plt.grid(True)
plt.show()
```

```python
# Bar Chart

months = ['Jan', 'Feb', 'Mar', 'April']
sales = [250, 300, 400, 350]

plt.bar(months, sales, color='skyblue')
plt.title('Monthly sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()
```

```python
plt.barh(months, sales, color='orange')
plt.title('Monthly sales (Horizontal)')
plt.xlabel('sales')
plt.ylabel('months')
plt.show()
```

```python
# Scatter Plot (Comparing Height vs Weight)

height = [150, 160, 170, 180]
weight = [50, 60, 65, 75]

plt.scatter(height, weight, color='green')
plt.title('Height vs Weight')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')

plt.show()
```

```python
# Pie Chart (Market Share)

import matplotlib.pyplot as plt

brands = ['Brand A', 'Brand B', 'Brand C']
shares = [45, 30, 25]

plt.pie(shares, labels=brands, autopct='%1.1f%%', startangle=90)

plt.title('Market Shares')
plt.axis('equal')
plt.show()
```

```python
# Multiple Plot in one Figure

x = [1, 2, 3, 4]
y1 = [1, 4, 9, 16]
y2 = [1, 2, 3, 4]

plt.plot(x, y1, label='Square')
plt.plot(x, y2, label='Linear')

plt.legend()
plt.title('Multiple Lines')
plt.show()
```

```python
# Subplots

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]

months = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [100, 150, 120, 200]

height = [150, 160, 170, 180, 190]
weight = [50, 55, 65, 75, 85]

share = [40, 30, 20, 10]
brands = ['A', 'B', 'C', 'D']

fig, axs = plt.subplots(2, 2)

axs[0, 0].plot(x, y1)
axs[0, 0].set_title('Square')

axs[0, 1].bar(months, sales)
axs[0, 1].set_title('Sales')

axs[1, 0].scatter(height, weight)
axs[1, 0].set_title('Scatter')

axs[1, 1].pie(share, labels=brands, autopct='%1.1f%%')
axs[1, 1].set_title('Pie Chart')

plt.tight_layout()
plt.show()
```

```python
# Subplots

fig, axs = plt.subplots(2, 2)

axs[0, 0].plot(x, y1)
axs[0, 0].set_title('Square')

axs[0, 1].bar(months, sales)
axs[0, 1].set_title('Sales')

axs[1, 0].scatter(height, weight)
axs[1, 0].set_title('Scatter')

axs[1, 1].pie(share, labels=brands, autopct='%1.1f%%')
axs[1, 1].set_title('Pie Chart')

plt.tight_layout()
plt.show()
```
